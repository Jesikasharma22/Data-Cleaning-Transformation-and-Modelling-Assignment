#1. What is Data Loading in Power BI and why is it considered the first step of analysis?

Data Loading is the process of importing data from different sources (Excel, CSV, SQL, etc.) into Power BI. It is the first step because data must be loaded before it can be cleaned, transformed, modeled, or visualized.

#2. Explain the difference between “Load” and “Transform Data” in Power BI.

| Load                                 | Transform Data                                                                             |
| ------------------------------------ | ------------------------------------------------------------------------------------------ |
| Imports data directly into Power BI. | Opens Power Query Editor to clean and modify data before loading.                          |
| No changes are made to the data.     | Data can be filtered, renamed, merged, split, remove duplicates, replace null values, etc. |
| Faster process.                      | Used for data cleaning and preparation.                                                    |


#3. What is a Fact Table and a Dimension Table? Give examples from the dataset.

**Fact Table:**

- Stores measurable values (numerical data).
- Example columns:
- Confirmed_Cases
- Recovered_Cases
- Deaths
- Active_Cases
- Tests_Conducted
- Hospitalized

**Dimension Table:**

- Stores descriptive information.
- Example columns:
- Country
- State
- Date
- Vaccination_Status

#4. Why is Star Schema preferred over Snowflake Schema in Power BI?

- Faster query performance.
- Simpler relationships.
- Easier to understand and maintain.
- Better report performance.
- Recommended by Microsoft for Power BI.

#5. Identify and remove duplicate records based on Date, Country, and State.

Power Query Steps:

1. Open Transform Data.
2. Select Date, Country, and State columns.
3. Go to Home → Remove Rows → Remove Duplicates.
4. Click Close & Apply.

#6. Identify and replace null values in Vaccination_Status.

Power Query Steps:

1. Select the Vaccination_Status column.
2. Click Transform → Replace Values.
3. Replace null with "Unknown" (or "Not Available").
4. Click OK.

#7. Create a new column to calculate Recovery Rate.

Formula (DAX):
```
Recovery Rate =
DIVIDE([Recovered_Cases], [Confirmed_Cases], 0) * 100
```
This returns the recovery percentage for each record.

#8. Create a summarized table showing total confirmed cases by Country.

DAX Formula:
```
Country Summary =
SUMMARIZE(
    'coronavirus_2020_dataset',
    'coronavirus_2020_dataset'[Country],
    "Total Confirmed Cases",
    SUM('coronavirus_2020_dataset'[Confirmed_Cases])
)
```
This creates a table showing each Country with its Total Confirmed Cases.
